```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("WEATHER Book2.csv") 
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]

df.head()

```

```python
df.isnull().sum()

# Fill missing values
df = df.ffill()
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce', dayfirst=True)

plt.plot(df['Date'], df['Temperature'])
plt.title("Temperature Over Time")
plt.xlabel("Date")
plt.ylabel("Temperature")
plt.show()
```

```python
df[['Temperature', 'Humidity']].corr()
```

```python
sns.heatmap(df[['Temperature', 'Humidity']].corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

###INSIGHTS###

-Temperature shows high fluctuations with no clear trend.

-Humidity values vary independently in the dataset.

-There is a very weak correlation (-0.023) between temperature and humidity

-No strong pattern or relationship is observed between the variables.
